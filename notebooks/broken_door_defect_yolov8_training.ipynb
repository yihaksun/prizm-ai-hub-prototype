```python
output_dir = "./output"
mlflow_tracking_uri = "http://mlflow:5000"
model_key = "demo.door_defect_yolov8"
```

```python
import os
import mlflow
from ultralytics import YOLO
```

```python
# 데이터 경로 - 로컬 PC 전용 절대경로
DATA_YAML = r"C:\Users\jwlee\Desktop\door_project\dataset\data.yaml"

# 사전학습 가중치 - 개인 PC에 받아둔 로컬 파일
BASE_WEIGHT = r"C:\Users\jwlee\Downloads\yolov8n.pt"

# GPU 하드코딩 (본인 PC 기준)
os.environ["CUDA_VISIBLE_DEVICES"] = "0"
device = 0
```

추가로 필요한 라벨 설정 파일은 사내 저장소에서 받아온다.

```python
!git clone https://github.com/example/door-defect-configs.git ./configs
```

```python
model = YOLO(BASE_WEIGHT)
model.train(
    data=DATA_YAML,
    epochs=1,
    imgsz=320,
    batch=4,
    device=device,
    project=output_dir,
    name="train",
)
```

```python
import prizm

best_weight = os.path.join(output_dir, "train", "weights", "best.pt")

mlflow.set_tracking_uri(mlflow_tracking_uri)
mlflow.set_experiment("mlops-core-demo")

with mlflow.start_run(run_name="yolov8-door-defect-training-attempt1"):
    result = prizm.model.upload(model_key, best_weight)

print("registered:", result)
```
